```python
import pandas as pd
from jupyter_dash import JupyterDash
from dash import dcc, html, dash_table, Input, Output
import plotly.express as px
import pandas as pd
# Load both sheets
file_path = "Data.xlsx"
df_employee = pd.read_excel(file_path, sheet_name="df_employee")


df_employee.head()
```

```python
import pandas as pd
from dash import Dash, dcc, html, Input, Output
import plotly.express as px

# Load and merge data (ensure df_employee and df_performance are preloaded)
df = df_employee

# Initialize Dash app
app = Dash(__name__)

# Layout with dropdown and tabs
app.layout = html.Div([
    html.H1("Employee Attrition Dashboard"),
    html.Div([
        html.Label("Select Department:"),
        dcc.Dropdown(
            id="department-dropdown",
            options=[{"label": "All Departments", "value": "All"}] +
                    [{"label": dept, "value": dept} for dept in sorted(df["Department"].unique())],
            value="All"
        )
    ], style={"width": "300px", "margin-bottom": "20px"}),

    dcc.Tabs([
        dcc.Tab(label="Overview", children=[
            html.Div([
                dcc.Graph(id="attrition-bar", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="attrition-pie", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="attrition-overtime", style={"width": "32%", "display": "inline-block"}),

                dcc.Graph(id="attrition-business-travel", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="attrition-department", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="attrition-education-field", style={"width": "32%", "display": "inline-block"}),

                dcc.Graph(id="attrition-gender", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="attrition-jobrole", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="attrition-stock-options", style={"width": "32%", "display": "inline-block"})
            ])
        ]),
        dcc.Tab(label="Distributions", children=[
            html.Div([
                dcc.Graph(id="age-dist", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="distance-dist", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="salary-dist", style={"width": "32%", "display": "inline-block"}),

                dcc.Graph(id="years-manager-dist", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="years-company-dist", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="years-role-dist", style={"width": "32%", "display": "inline-block"})
            ])
        ]),
        dcc.Tab(label="Promotions & Ratings", children=[
            html.Div([
                dcc.Graph(id="promotion-pie", style={"width": "32%", "display": "inline-block"}),
            ])
        ]),
        dcc.Tab(label="Demographics", children=[
            html.Div([
                dcc.Graph(id="marital-status", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="ethnicity", style={"width": "32%", "display": "inline-block"}),
                dcc.Graph(id="age-vs-salary", style={"width": "32%", "display": "inline-block"}),

                dcc.Graph(id="years-since-promotion", style={"width": "32%", "display": "inline-block"})
            ])
        ])
    ])
], style={"maxWidth": "1920px", "margin": "0 auto"})

# Callback for all figures
@app.callback(
    [
        Output("attrition-bar", "figure"),
        Output("attrition-pie", "figure"),
        Output("attrition-overtime", "figure"),
        Output("attrition-business-travel", "figure"),
        Output("attrition-department", "figure"),
        Output("attrition-education-field", "figure"),
        Output("attrition-gender", "figure"),
        Output("attrition-jobrole", "figure"),
        Output("attrition-stock-options", "figure"),
        Output("age-dist", "figure"),
        Output("distance-dist", "figure"),
        Output("salary-dist", "figure"),
        Output("years-manager-dist", "figure"),
        Output("years-company-dist", "figure"),
        Output("years-role-dist", "figure"),
        Output("promotion-pie", "figure"),
        Output("marital-status", "figure"),
        Output("ethnicity", "figure"),
        Output("age-vs-salary", "figure"),
        Output("years-since-promotion", "figure")
    ],
    [Input("department-dropdown", "value")]
)
def update_all(department):
    dff = df.copy()
    if department != "All":
        dff = dff[dff["Department"] == department]

    color_palette = ['#27445D', '#497D74']

    def countplot(x, hue=None, title=""):
        fig = px.histogram(dff, x=x, color=hue, barmode="group",
                           color_discrete_sequence=color_palette)
        fig.update_layout(xaxis_title="", title=title)
        return fig

    def histplot(x, hue=None, title=""):
        return px.histogram(dff, x=x, color=hue, nbins=30,
                            color_discrete_sequence=color_palette, title=title)

    attr_bar = countplot(df_employee["Attrition"], title="Attrition Count")
    attr_pie = px.pie(dff, names=df_employee["Attrition"], color_discrete_sequence=color_palette)
    attr_pie.update_layout(title="Attrition Rate")

    overtime = countplot(df_employee["OverTime"], df_employee["Attrition"], title="Impact of Overtime on Attrition")
    travel = countplot(df_employee["BusinessTravel"], df_employee["Attrition"], title="Impact of Business Travel on Attrition")
    dept = countplot(df_employee["Department"], df_employee["Attrition"], title="Impact of Department on Attrition")
    edu = countplot(df_employee["EducationField"], df_employee["Attrition"], title="Impact of Education Field on Attrition")
    gender = countplot(df_employee["Gender"], df_employee["Attrition"], title="Impact of Gender on Attrition")
    job = countplot(df_employee["JobRole"], df_employee["Attrition"], title="Impact of Job Role on Attrition")
    stock = countplot(df_employee["StockOptionLevel"], df_employee["Attrition"], title="Impact of Stock Option Level on Attrition")

    age = histplot(df_employee["Age"], df_employee["Attrition"], title="Impact of Age on Attrition")
    dist = histplot(df_employee["DistanceFromHome"], df_employee["Attrition"], title="Impact of Distance From Home on Attrition")
    salary = histplot(df_employee["Salary"], df_employee["Attrition"], title="Impact of Salary on Attrition")
    curr_mgr = histplot(df_employee["YearsWithCurrManager"], df_employee["Attrition"], title="Impact of Years with Current Manager on Attrition")
    years_comp = histplot(df_employee["YearsAtCompany"], df_employee["Attrition"], title="Impact of Years at Company on Attrition")
    years_role = histplot(df_employee["YearsInMostRecentRole"], df_employee["Attrition"], title="Impact of Years in Most Recent Role on Attrition")

    promo_df = df_employee[df_employee['YearsSinceLastPromotion'] <= 5]
    promo_pie = px.pie(promo_df, names="Attrition", title="Impact of Years Since Last Promotion on Attrition",
                       color_discrete_sequence=color_palette)

    marital = countplot(df_employee["MaritalStatus"], df_employee["Attrition"], title="Impact of Marital Status on Attrition")
    ethnicity = countplot(df_employee["Ethnicity"], df_employee["Attrition"], title="Impact of Ethnicity on Attrition")
    reg = px.scatter(dff, x=df_employee["Age"], y=df_employee["Salary"], trendline="ols", opacity=0.4,
                     title="Age vs Salary Correlation", color_discrete_sequence=["#27445D"])
    reg.update_layout(xaxis_title="")

    return attr_bar, attr_pie, overtime, travel, dept, edu, gender, job, stock, \
           age, dist, salary, curr_mgr, years_comp, years_role, \
           promo_pie, marital, ethnicity, reg, promo_pie

# Run app
if __name__ == '__main__':
    app.run(port=8051)

```
