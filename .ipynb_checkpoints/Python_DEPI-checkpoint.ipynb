# EDA(Exploratory Data Analysis)

```python
Attrition_Count = sns.countplot(x=emp.Attrition)

for p in Attrition_Count.patches:
    height = p.get_height()
    Attrition_Count.annotate(f'{height}', 
                (p.get_x() + p.get_width() / 2., height), 
                ha='center', va='bottom', 
                fontsize=11)

plt.show()
```

```python
attrition_counts = emp['Attrition'].value_counts()
plt.pie(attrition_counts, labels=attrition_counts.index, autopct='%1.1f%%', colors=['#27445D','#497D74'])
plt.title('Attrition Rate')
plt.show()
```

```python
sns.countplot(x=emp.State , hue=emp.Attrition , palette="crest")
plt.show()
```

# 14.Impact of years since last promotion on attrition

```python
promotion_stagnation = emp[emp['YearsSinceLastPromotion'] <= 5]['Attrition'].value_counts(normalize=True) * 100

plt.pie(promotion_stagnation , labels=attrition_counts.index , autopct='%1.1f%%' , colors=['#27445D','#497D74'])
plt.title('Attrition rate for employees not promoted in less than 5 years')
plt.show()
```

```python
promotion_stagnation = emp[emp['YearsSinceLastPromotion'] >= 5]['Attrition'].value_counts(normalize=True) * 100

plt.pie(promotion_stagnation , labels=attrition_counts.index , autopct='%1.1f%%' , colors=['#27445D','#497D74'])
plt.title('Attrition rate for employees promoted in less than 5 years')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.regplot(x='Age', y='Salary', data=emp, scatter_kws={'alpha':0.3})
plt.title('Age vs Salary (with Regression Line)')
plt.show()
```
